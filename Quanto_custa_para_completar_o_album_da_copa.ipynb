#Importar os pacotes

```python
import random
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
#Informações do Album da Copa 2026

valor_pacote = 7
valor_album = 24.90 #consulta no site da panini 04-06-2026 capa cartao
qtde_figurinhas_por_pacote = 7
qtde_figurinhas_do_album = 980
```

#Analise por quantidade de Albuns sem figurinhas repetidas no pacote

```python
def completa_album(qtde_figurinhas_do_album, qtde_figurinhas_por_pacote):

    album = set()
    qtde_de_compras_pacote_sem_repetida = 0
    while len(album) < qtde_figurinhas_do_album:
        pacote_sem_repeticao = set(random.sample(range(1, qtde_figurinhas_do_album+1), qtde_figurinhas_por_pacote))
        album = album.union(pacote_sem_repeticao)
        qtde_de_compras_pacote_sem_repetida += 1

    return qtde_de_compras_pacote_sem_repetida
```

```python
n = 5000
albuns_sem_repeticao = [completa_album(980,7) for i in range(n)]
```

```python
#criar um histograma
plt.figure(figsize=(10,6))
plt.hist(albuns_sem_repeticao, color= 'blue')
plt.title('Histograma do número de compras necessárias para completar o álbum')
plt.grid()
plt.xlabel('Número de pacotes')
plt.ylabel('Qtde Albuns')
plt.show()
```

```python
#custo para completar o album
mediana = pd.Series(albuns_sem_repeticao).median()
custo_album = mediana * valor_pacote
custo_album_total = custo_album + valor_album
print(f'A mediana é {mediana}')
print(f'O custo para completar o álbum é {custo_album}')
print(f'O Valor total para completar o album é {custo_album_total}')
```

#Analise por quantidade de Albuns com possibilidade de figurinha repetida

```python
def completa_album_com_repeticao(qtde_figurinhas_do_album, qtde_figurinhas_por_pacote):
    album2 = set()
    qtde_de_compras_pacote = 0
    while len(album2) < qtde_figurinhas_do_album:
        pacote = {random.randint(1,qtde_figurinhas_do_album) for i in range(qtde_figurinhas_por_pacote)}
        album2 = album2.union(pacote)
        qtde_de_compras_pacote += 1

    return qtde_de_compras_pacote
```

```python
n = 5000
albuns_com_repeticao = [completa_album_com_repeticao(980,7) for i in range(n)]
```

```python
#criar um histograma
plt.figure(figsize=(10,6))
plt.hist(albuns_com_repeticao, color= 'red')
plt.title('Histograma do número de compras necessárias para completar o álbum')
plt.grid()
plt.xlabel('Número de pacotes')
plt.ylabel('Qtde Albuns')
plt.show()
```

```python
#custo para completar o album
mediana = pd.Series(albuns_com_repeticao).median()
custo_album = mediana * valor_pacote
custo_album_total = custo_album + valor_album
print(f'A mediana é {mediana}')
print(f'O custo para completar o álbum é {custo_album}')
print(f'O Valor total para completar o album é {custo_album_total}')
```
